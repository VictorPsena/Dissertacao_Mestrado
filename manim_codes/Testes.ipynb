```python
%%manim -v WARNING -qm slide1

config.media_embed = True
config.media_width = "100%"


Nomes_Intruducao = {
    "autor": "Victor Sena",
    "orientadora": "Rosana Sueli da Motta Jafelice ",
    "coorientadora": "Caio August Rodrigues dos Santos "
}

class slide1(Slide):
    def construct(self):
        nome1 = Nomes_Intruducao["autor"]
        nome2 = Nomes_Intruducao["orientadora"]
        nome3 = Nomes_Intruducao["coorientadora"]
        self.play(Write(Text(nome1)))
        self.next_slide()
        self.play(Write(Text(nome2)))
        self.next_slide()
        self.play(Write(Text(nome3)))
```
